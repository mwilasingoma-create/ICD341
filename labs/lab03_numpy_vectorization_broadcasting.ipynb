1. What is vectorization?

Vectorization is the process of performing an operation on an entire NumPy array without explicitly writing a Python loop for each element.

2. Why is vectorization useful?

Vectorization makes code shorter, clearer, easier to maintain, and generally faster, especially when processing large datasets.

3. Difference between a Python loop and a NumPy vectorized operation?

A Python loop processes elements one at a time using explicit iteration, while a NumPy vectorized operation performs the operation on the entire array using optimized numerical operations.

```python
marks = np.array([40, 55, 62, 70, 81, 90])

print("Original marks:", marks)
```

```python
adjusted = marks + 5

print("Adjusted marks:", adjusted)
```

```python
percentage = marks / 100

print("Percentage:", percentage)
```

```python
z = (marks - np.mean(marks)) / np.std(marks)

print("Standardized values:", z)
print("Mean of z:", np.mean(z))
```

```python
data = np.array([
    [60, 70, 80],
    [50, 65, 75],
    [90, 85, 95]
])

adjustment = np.array([5, 0, 2])

print("Data:")
print(data)

print("Adjustment:")
print(adjustment)
```

```python
result = data + adjustment

print("Result:")
print(result)
```

Broadcasting allows NumPy to perform operations between arrays with compatible shapes. In this example, the 1 × 3 adjustment array is automatically applied to each row of the 3 × 3 data array. NumPy does this without explicitly copying the adjustment array.

```python
data = np.array([
    [10, 20, 30],
    [20, 30, 40],
    [30, 40, 50]
])

print("Data:")
print(data)
```

```python
means = data.mean(axis=0)

print("Column means:", means)
```

```python
centered = data - means

print("Centered data:")
print(centered)
```

```python
print("Mean of centered columns:", centered.mean(axis=0))
```

The vector of column means has shape (3,), while the data matrix has shape (3, 3). NumPy broadcasts the three means across the corresponding columns and subtracts each mean from every value in that column. Therefore, each centered column has a mean of approximately zero.

```python
x = np.arange(1_000_000)

print("Number of values:", len(x))
print("First 10 values:", x[:10])
```

```python
result_loop = []

for value in x:
    result_loop.append(value**2 + 2*value + 1)

print("First 10 loop results:", result_loop[:10])
```

```python
result_vectorized = x**2 + 2*x + 1

print("First 10 vectorized results:", result_vectorized[:10])
```

```python
print("Results equal:", np.allclose(result_loop, result_vectorized))
```

1. Are the two results equal?

Yes. np.allclose() returns True, showing that the loop-based and vectorized results agree.

2. Which approach is easier to read?

The vectorized NumPy approach is generally easier to read because the mathematical operation is expressed directly without an explicit loop.

3. Which approach would you expect to be faster?

The NumPy vectorized approach would generally be faster for a large array.

4. Why?

NumPy performs vectorized operations using optimized numerical routines implemented largely in compiled code, reducing the overhead of executing a Python loop for each element.

```python
np.random.seed(123)

x = np.random.normal(100, 15, 100_000)

print("Number of observations:", len(x))
print("First 10 observations:", x[:10])
```

```python
z = (x - x.mean()) / x.std()

print("Mean of z:", z.mean())
print("Standard deviation of z:", z.std())
```

Vectorization is particularly useful for large statistical datasets because it allows mathematical operations to be performed on many observations at once without explicitly writing Python loops. This results in shorter and clearer code and generally improves computational efficiency. In this example, 100,000 observations were standardized using a single vectorized expression.

```python
def standardize(x):
    return (x - np.mean(x)) / np.std(x)
```

```python
x1 = np.array([10, 20, 30, 40, 50])
x2 = np.array([5, 10, 15, 20, 25])
x3 = np.array([100, 150, 200, 250, 300])

z1 = standardize(x1)
z2 = standardize(x2)
z3 = standardize(x3)

print("Standardized x1:", z1)
print("Mean of x1:", np.mean(z1))
print()

print("Standardized x2:", z2)
print("Mean of x2:", np.mean(z2))
print()

print("Standardized x3:", z3)
print("Mean of x3:", np.mean(z3))
```

```python
def standardize_columns(x):
    return (x - np.mean(x, axis=0)) / np.std(x, axis=0)
```

```python
data_2d = np.array([
    [10, 20, 30],
    [20, 30, 40],
    [30, 40, 50],
    [40, 50, 60]
])

standardized_data = standardize_columns(data_2d)

print("Original data:")
print(data_2d)

print("\nStandardized data:")
print(standardized_data)

print("\nColumn means:")
print(standardized_data.mean(axis=0))

print("\nColumn standard deviations:")
print(standardized_data.std(axis=0))
```

For a 2D dataset:
axis=0 calculates statistics down each column. Therefore, np.mean(x, axis=0) produces one mean for every column, while np.std(x, axis=0) produces one standard deviation for every column. NumPy then broadcasts these vectors across the rows, allowing every column to be standardized simultaneously.
