## 1. Creating and Inspecting NumPy Arrays

- Number of dimensions.

```python
x = np.array([10, 20, 30, 40, 50])

print(x)
print(x.shape)
print(x.ndim)
print(x.dtype)
```

## 2. Creating Arrays Using NumPy Functions

NumPy provides several functions for generating arrays efficiently.

- `np.zeros()` creates an array containing zeros.
- `np.ones()` creates an array containing ones.
- `np.arange()` generates values within a specified range.
- `np.linspace()` generates evenly spaced values between two endpoints.

```python
a = np.zeros(10)
b = np.ones(10)
c = np.arange(1, 11)
d = np.linspace(0, 1, 11)

print("a = - lab02_numpy_fundamentals.ipynb:6", a)
print("b = - lab02_numpy_fundamentals.ipynb:7", b)
print("c = - lab02_numpy_fundamentals.ipynb:8", c)
print("d = - lab02_numpy_fundamentals.ipynb:9", d)
```

What does np.zeros() do?
It creates a NumPy array containing the specified number of zeros.
What does np.ones() do?
It creates a NumPy array containing the specified number of ones.
Difference between arange() and linspace()?
np.arange() generates values using a specified step size, while np.linspace() generates a specified number of evenly spaced values between a start and end point.
Statistical applications:
np.zeros() — initializing an array to store statistical results.
np.ones() — creating weights or baseline values.
np.arange() — generating observation numbers or discrete sequences.
np.linspace() — generating evenly spaced values for simulations and statistical graphs.

## 3. Indexing and Slicing Arrays

Indexing is used to access individual elements of an array, while slicing is used to access a range or pattern of elements.

NumPy uses zero-based indexing, meaning the first element has index `0`.

The following examples demonstrate:
- Accessing the first and last elements.
- Selecting the first three elements.
- Selecting a range of elements.
- Selecting every second element.

```python
marks = np.array([45, 67, 82, 56, 91, 73, 64, 88])

print("First mark: - lab02_numpy_fundamentals.ipynb:3", marks[0])
print("Last mark: - lab02_numpy_fundamentals.ipynb:4", marks[-1])
print("First three marks: - lab02_numpy_fundamentals.ipynb:5", marks[:3])
print("Marks from position 3 to 6: - lab02_numpy_fundamentals.ipynb:6", marks[2:6])
print("Every second mark: - lab02_numpy_fundamentals.ipynb:7", marks[::2])
```

## 4. Statistical Operations

NumPy provides built-in functions for common statistical calculations.

The following operations calculate the:
- Mean.
- Median.
- Standard deviation.
- Minimum.
- Maximum.
- Sum.

These functions are useful when performing statistical analysis on numerical datasets.

```python
print("Mean: - lab02_numpy_fundamentals.ipynb:1", np.mean(marks))
print("Median: - lab02_numpy_fundamentals.ipynb:2", np.median(marks))
print("Standard deviation: - lab02_numpy_fundamentals.ipynb:3", np.std(marks))
print("Minimum: - lab02_numpy_fundamentals.ipynb:4", np.min(marks))
print("Maximum: - lab02_numpy_fundamentals.ipynb:5", np.max(marks))
print("Sum: - lab02_numpy_fundamentals.ipynb:6", np.sum(marks))
```

Interpretation:
The mean mark is 70.75, representing the average mark obtained by the students. The median is 70, meaning that the middle mark is 70 when the marks are arranged in order. The standard deviation is approximately 14.91, indicating the amount of variation or spread of the marks around the mean. The minimum mark is 45, while the maximum mark is 91, showing that the marks range from 45 to 91. The total of all the marks is 566.

## 5. Two-Dimensional NumPy Arrays

A two-dimensional NumPy array can represent data organized into rows and columns.

In this example, the dataset represents marks arranged in a matrix.

The `shape` property shows the number of rows and columns, while `ndim` shows the number of dimensions.

```python
data = np.array([
    [45, 67, 82],
    [56, 91, 73],
    [64, 88, 76]
])

print(data)
print(data.shape)
print(data.ndim)
```

## 6. Indexing and Slicing Two-Dimensional Arrays

Two-dimensional arrays can be accessed using row and column indices.

The examples demonstrate how to:
- Select a complete row.
- Select a complete column.
- Access an individual value.
- Select multiple rows.
- Select multiple columns.

The general indexing format is:

`array[row, column]`

```python
print("First row: - lab02_numpy_fundamentals.ipynb:1", data[0])
print("Second column: - lab02_numpy_fundamentals.ipynb:2", data[:, 1])
print("Value 91: - lab02_numpy_fundamentals.ipynb:3", data[1, 1])
print("First two rows: - lab02_numpy_fundamentals.ipynb:4")
print(data[:2])
print("Last two columns: - lab02_numpy_fundamentals.ipynb:6")
print(data[:, 1:])
```

## 7. Calculating Column Means

The mean can be calculated across columns using `axis=0`.

This produces one mean value for each subject/column.

Using `axis=0` means that the calculation proceeds down the rows while producing a result for each column.

```python
subject_means = np.mean(data, axis=0)

print("Mean mark for each subject: - lab02_numpy_fundamentals.ipynb:3", subject_means)
```

## 8. Calculating Row Means

The mean can be calculated across rows using `axis=1`.

This produces one mean value for each student/row.

Using `axis=1` means that the calculation is performed across the columns for each row.

```python
student_means = np.mean(data, axis=1)

print("Mean mark for each student: - lab02_numpy_fundamentals.ipynb:3", student_means)
```

Axis explanation:
axis=0 calculates the operation down the rows, producing results for each column. Therefore, np.mean(data, axis=0) calculates the mean mark for each subject. axis=1 calculates the operation across the columns, producing results for each row. Therefore, np.mean(data, axis=1) calculates the mean mark for each student.

## 9. Simulating Statistical Data

NumPy can generate random data from statistical distributions.

The `np.random.normal()` function generates observations from a normal distribution.

In this example:
- `loc=50` specifies the mean.
- `scale=10` specifies the standard deviation.
- `size=1000` generates 1,000 observations.

This is useful for demonstrating statistical analysis using simulated datasets.

```python
x = np.random.normal(loc=50, scale=10, size=1000)

print("First 10 observations: - lab02_numpy_fundamentals.ipynb:3", x[:10])
print("Number of observations: - lab02_numpy_fundamentals.ipynb:4", len(x))
```

## 10. Summary Statistics of Simulated Data

The simulated dataset is summarized using common descriptive statistics.

These statistics provide information about the centre and spread of the generated observations, as well as their minimum and maximum values.

```python
print("Mean: - lab02_numpy_fundamentals.ipynb:1", np.mean(x))
print("Median: - lab02_numpy_fundamentals.ipynb:2", np.median(x))
print("Standard deviation: - lab02_numpy_fundamentals.ipynb:3", np.std(x))
print("Minimum: - lab02_numpy_fundamentals.ipynb:4", np.min(x))
print("Maximum: - lab02_numpy_fundamentals.ipynb:5", np.max(x))
```

## 11. Data Visualization

A histogram is used to visualize the distribution of the simulated observations.

The x-axis represents the observed values, while the y-axis represents their frequency.

Histograms are useful for examining the shape and distribution of numerical data.

```python
import matplotlib.pyplot as plt

plt.hist(x, bins=30)
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Distribution of Simulated Data")
plt.show()
```
