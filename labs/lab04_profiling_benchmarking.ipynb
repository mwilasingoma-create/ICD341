## Part A — `%timeit`

`%timeit` is a Jupyter/IPython magic command used to measure the execution time of Python code.

It runs the expression multiple times and reports statistics about the execution time. This provides a more reliable performance measurement than timing a single execution.

The following example benchmarks a NumPy vectorized operation on one million values.

````python
x = np.arange(1_000_000)

%timeit x ** 2
````

## Part B — Comparing a Python Loop with NumPy Vectorization

This section compares two implementations of the same mathematical operation.

The first implementation uses a traditional Python `for` loop, while the second uses NumPy vectorization.

Both functions calculate the square of every element in an array. `%timeit` is used to measure their execution times on the current computer.

````python
def square_loop(x):
    result = []
    for value in x:
        result.append(value ** 2)
    return result
````

````python
%timeit square_loop(x)
````

````python
def square_numpy(x):
    return x ** 2
````

````python
%timeit square_numpy(x)
````

### Task 1 — Benchmark Comparison

| Method | Measured Time |
|---|---:|
| Python loop | 145 ms ± 8.89 ms |
| NumPy vectorization | 3.57 ms ± 546 μs |

#### Questions

**1. Which implementation was faster?**

The NumPy vectorized implementation was faster on this computer. Its measured mean execution time was 3.57 ms compared with 145 ms for the Python loop.

**2. How many times faster was it?**

Using the mean benchmark times:

Speed-up = Loop time / NumPy time

Speed-up = 145 / 3.57 ≈ **40.6×**

Therefore, the NumPy implementation was approximately 40.6 times faster in this benchmark.

**3. Why might NumPy be faster?**

NumPy performs array operations using optimized numerical routines implemented largely in compiled code. This reduces the overhead associated with executing a Python `for` loop for each individual element.

The exact speed-up depends on the computer, Python version, NumPy version, array size, and system conditions. Therefore, benchmark results should be measured rather than assumed.

## Part F — Identifying an Inefficient Function

This section investigates a deliberately inefficient function that calculates the sum of squared values using a Python loop.

The function is profiled to identify which operations consume the most execution time. It is then compared with a vectorized NumPy implementation.

````python
def slow_function(x):
    result = []
    for value in x:
        result.append(value ** 2)
    total = sum(result)
    return total
````

````python
x = np.arange(1_000_000)
````

````python
%lprun -f slow_function slow_function(x)
````

````python
def fast_function(x):
    return np.sum(x ** 2)
````

````python
%timeit slow_function(x)
%timeit fast_function(x)
````

## Task 4 — Comparing the Inefficient and Optimized Functions

The deliberately inefficient `slow_function()` uses a Python `for` loop to square each element individually, stores the results in a Python list, and then calculates the total using `sum()`.

The optimized `fast_function()` uses NumPy vectorization and `np.sum()` to perform the same calculation.

### Benchmark Results

| Function | Measured Time |
|---|---:|
| `slow_function(x)` | 222 ms ± 47.1 ms |
| `fast_function(x)` | 4.56 ms ± 488 μs |

### Speed-up

Speed-up = Slow function time / Fast function time

Speed-up = 222 / 4.56 ≈ **48.7×**

Therefore, the optimized NumPy implementation was approximately **48.7 times faster** than the Python-loop implementation in this benchmark.

### Why is `fast_function()` more efficient?

The `slow_function()` performs element-by-element operations in a Python `for` loop and repeatedly appends values to a Python list. These operations introduce significant Python-level overhead.

The `fast_function()` uses NumPy vectorization:

## Part G — Mini Statistical Computing Investigation

This section compares two approaches for calculating the sum of squared deviations from the mean.

The first approach uses a Python `for` loop, while the second uses NumPy vectorization.

The experiment uses a large dataset containing 5,000,000 normally distributed observations. The two implementations will be benchmarked to determine the performance difference between loop-based and vectorized statistical computation.

````python
np.random.seed(42)

data = np.random.normal(
    loc=100,
    scale=20,
    size=5_000_000
)
````

````python
def squared_deviations_loop(x):
    mean = np.mean(x)
    total = 0
    for value in x:
        total += (value - mean) ** 2
    return total
````

````python
def squared_deviations_numpy(x):
    mean = np.mean(x)
    return np.sum((x - mean) ** 2)
````

````python
%timeit -n 1 -r 3 squared_deviations_loop(data)
````

````python
%timeit -n 1 -r 3 squared_deviations_numpy(data)
````

### Part G — Results and Analysis

The two implementations were benchmarked using the same dataset containing 5,000,000 normally distributed observations.

| Approach | Measured Time |
|---|---:|
| Python loop | 1.21 s ± 82.3 ms |
| NumPy vectorization | 44.2 ms ± 2.52 ms |

#### 1. Which approach was faster?

The NumPy vectorized approach had the lower measured execution time in this benchmark.

#### 2. What was the measured time for each approach?

- Python loop: **1.21 seconds ± 82.3 ms**
- NumPy vectorization: **44.2 ms ± 2.52 ms**

#### 3. Speed-up

Speed-up = Loop time / NumPy time

Speed-up = 1.21 / 0.0442 ≈ **27.4×**

Therefore, the NumPy implementation was approximately **27.4 times faster** in this benchmark.

#### 4. Why can vectorization produce a large performance difference?

The loop-based implementation performs the calculation one element at a time using Python-level iteration. This introduces overhead for processing each individual element.

The NumPy implementation performs the array operation using optimized numerical routines operating on the entire array. This reduces Python-level overhead and makes better use of optimized compiled code.

The large difference becomes particularly noticeable when processing millions of observations.

#### 5. How does profiling help decide where to optimize a statistical program?

Profiling identifies which parts of a program consume the most execution time. This allows optimization efforts to focus on actual computational bottlenecks rather than making unnecessary changes to code that has little effect on overall performance.

In this laboratory, profiling showed that the loop and list operations were major contributors to the execution time of `slow_function()`. Benchmarking then demonstrated the performance improvement obtained by replacing the loop-based approach with NumPy vectorization.
